## What Is Extractive Question Answering?

In extractive question answering, the model receives:

```text
Question
+
Context

and returns an answer span from the context.

Example:

Question:
What is BERT?

Context:
BERT is an encoder-only Transformer model.

Answer:
encoder-only Transformer model

The model does not freely generate an answer.

It selects the most likely answer span from the given context.

# Section 3: Building a Reusable Manual Extractive QnA Function

In the previous section, we manually performed one QnA forward pass.

```text
Question + Context

## Why We Need a Reusable Function

In a real project, we will not manually write the same code again and again.

We need a function that accepts:

```text
question

tokenizer

and returns:
```text

## Important Improvement

In the previous section, we used:

```python
torch.argmax(start_logits)
torch.argmax(end_logits)

independently.

This can create a problem.

Example:

start_index = 40

end_index = 25

This is invalid because the answer cannot end before it starts.

So we will build a more reliable span selector:

Choose the best valid span where:

start_index <= end_index

## Span Selection Idea

The model gives:

```text
start_logits: score for each token being answer start
end_logits:   score for each token being answer end

We will search for a good pair:
```text

(start_index, end_index)

such that:

start_index <= end_index

and the answer span is not too long.

This is more stable than independent argmax.

`````python
# ============================================================
# Cell 19: Helper Function to Select Best Valid Answer Span
# ============================================================

def select_best_answer_span(
    start_logits,
    end_logits,
    max_answer_length=40
):
    """
    Select the best valid answer span.

    Instead of independently selecting:
        argmax(start_logits)
        argmax(end_logits)

    we search over valid spans where:
        start_index <= end_index
        answer length <= max_answer_length

    Score of a span:
        start_logit + end_logit
    """

    start_logits = start_logits[0]
    end_logits = end_logits[0]

    best_score = -float("inf")
    best_start = 0
    best_end = 0

    sequence_length = start_logits.shape[0]

    for start_index in range(sequence_length):
        max_end_index = min(
            start_index + max_answer_length,
            sequence_length
        )

        for end_index in range(start_index, max_end_index):
            span_score = (
                start_logits[start_index].item()
                + end_logits[end_index].item()
            )

            if span_score > best_score:
                best_score = span_score
                best_start = start_index
                best_end = end_index

    return best_start, best_end, best_score
`````

## Observation: Span Selection

This function searches for a valid answer span.

It avoids impossible cases like:

```text
end token before start token

it also avoids extremely long answer spans using:

max_answer_length

This makes the QnA function more reliable for document chunks.

`````python
# ============================================================
# Cell 20: Reusable Manual Extractive QnA Function
# ============================================================

def manual_extractive_qa(
    question,
    context,
    tokenizer,
    model,
    device=device,
    max_length=384,
    max_answer_length=40
):
    """
    Manual extractive question answering.

    Input:
        question: user question
        context: text passage where answer may exist
        tokenizer: Hugging Face tokenizer
        model: Hugging Face QnA model

    Output:
        dictionary containing answer and useful metadata
    """

    encoded = tokenizer(
        question,
        context,
        return_tensors="pt",
        truncation=True,
        max_length=max_length
    )

    encoded = {
        key: value.to(device)
        for key, value in encoded.items()
    }

    with torch.no_grad():
        outputs = model(**encoded)

    start_logits = outputs.start_logits
    end_logits = outputs.end_logits

    start_index, end_index, span_score = select_best_answer_span(
        start_logits=start_logits,
        end_logits=end_logits,
        max_answer_length=max_answer_length
    )

    input_ids = encoded["input_ids"][0]

    answer_token_ids = input_ids[start_index : end_index + 1]

    answer = tokenizer.decode(
        answer_token_ids,
        skip_special_tokens=True
    ).strip()

    start_probability = torch.softmax(start_logits, dim=-1)[0, start_index].item()
    end_probability = torch.softmax(end_logits, dim=-1)[0, end_index].item()

    confidence_score = (start_probability + end_probability) / 2

    result = {
        "question": question,
        "answer": answer,
        "confidence_score": round(confidence_score, 4),
        "span_score": round(span_score, 4),
        "start_token_index": start_index,
        "end_token_index": end_index,
        "context_length_characters": len(context)
    }

    return result
`````

`````python
# ============================================================
# Cell 27: Define Basic Text Cleaning Function
# ============================================================

import re

def clean_text(text):
    """
    Basic cleaning for PDF-extracted text.

    This removes repeated spaces, newlines, and extra whitespace.
    """

    text = re.sub(r"\s+", " ", text)
    text = text.strip()

    return text
`````

## Why Text Cleaning Is Needed

```text

For a QnA model, cleaner text usually gives better results.

This cleaning function is intentionally simple.

Later, we can improve it if needed.

`````python
# ============================================================
# Cell 28: Extract PDF Text Page-Wise Using PyMuPDF
# ============================================================

import fitz

def extract_pdf_pagewise(pdf_path):
    """
    Extract text from every page of a PDF.

    Output:
        List of dictionaries:
        [
            {
                "page_number": 1,
                "text": "..."
            },
            ...
        ]
    """

    doc = fitz.open(pdf_path)

    pages = []

    for page_index, page in enumerate(doc):
        page_number = page_index + 1
        raw_text = page.get_text()
        cleaned_page_text = clean_text(raw_text)

        pages.append({
            "page_number": page_number,
            "text": cleaned_page_text,
            "num_characters": len(cleaned_page_text)
        })

    doc.close()

    return pages
`````

# Section 5: Chunking Long PDF Text for Document QnA

In the previous section, we extracted PDF text page-wise.

We also tried QnA on selected pages.

But we saw one important limitation:

```text

The model has a maximum input length.

Full PDF pages
→ Smaller chunks
→ QnA on each chunk
→ Rank candidate answers

This is the first real step toward RAG.

## Why Chunking Is Needed

Our QnA model receives:

```text
question + context


But the tokenizer has a limit.

For example, in our function we used:


max_length = 384

If the context is longer than this, the tokenizer truncates it.

That means:

The model may never see the part of the document that contains the answer.

Chunking solves this by breaking the document into smaller pieces.

## Chunking Idea

Instead of passing the full document:

```text
Page 1 + Page 2 + Page 3 + ... + Page N

We create smaller chunks:
```text
Chunk 0: Page 1, words 0 to 180
Chunk 1: Page 1, words 140 to 320
Chunk 2: Page 1, words 280 to 460
Chunk 3: Page 2, words 0 to 180
...

Notice the overlap.

Overlap helps because an answer may lie near the boundary between two chunks.

## Why Overlap Is Useful

Without overlap:

```text
Chunk 1 ends here | Chunk 2 starts here

If the answer is split across the boundary, the model may miss it.

With overlap:

Chunk 1: words 0 to 180
Chunk 2: words 140 to 320

Some words are repeated.

This gives the model another chance to see the full answer.

`````python
# ============================================================
# Cell 37: Chunk Page-Wise PDF Text by Words
# ============================================================

def chunk_pages_by_words(
    pages,
    chunk_size=180,
    overlap=40,
    min_chunk_words=30
):
    """
    Convert page-wise PDF text into overlapping word chunks.

    Each chunk keeps useful metadata:
        chunk_id
        page_number
        start_word_index
        end_word_index
        text
        num_words
        num_characters

    Parameters:
        pages:
            List of page dictionaries from extract_pdf_pagewise()

        chunk_size:
            Number of words in each chunk

        overlap:
            Number of words repeated between consecutive chunks

        min_chunk_words:
            Ignore very small chunks below this length
    """

    if overlap >= chunk_size:
        raise ValueError("overlap must be smaller than chunk_size")

    chunks = []
    chunk_id = 0

    for page in pages:
        page_number = page["page_number"]
        page_text = page["text"]

        words = page_text.split()

        start = 0

        while start < len(words):
            end = min(start + chunk_size, len(words))

            chunk_words = words[start:end]
            chunk_text = " ".join(chunk_words)
            chunk_text = clean_text(chunk_text)

            if len(chunk_words) >= min_chunk_words:
                chunks.append({
                    "chunk_id": chunk_id,
                    "page_number": page_number,
                    "start_word_index": start,
                    "end_word_index": end,
                    "text": chunk_text,
                    "num_words": len(chunk_words),
                    "num_characters": len(chunk_text)
                })

                chunk_id += 1

            if end == len(words):
                break

            start = end - overlap

    return chunks
`````

## Observation: Chunk Function

This function does not destroy page information.

Each chunk still knows:

```text
which page it came from
where it starts
where it ends

This is useful for:

source tracking
debugging
answer verification
future RAG
fine-tuning dataset preparation

`````python
# ============================================================
# Cell 38: Create Chunks from the Uploaded PDF
# ============================================================

chunks = chunk_pages_by_words(
    pages=pages,
    chunk_size=180,
    overlap=40,
    min_chunk_words=30
)

print("Total pages:", len(pages))
print("Total chunks created:", len(chunks))

print("\nFirst 3 chunks:")
for chunk in chunks[:3]:
    print("Chunk ID:", chunk["chunk_id"])
    print("Page:", chunk["page_number"])
    print("Words:", chunk["num_words"])
    print("Characters:", chunk["num_characters"])
    print("Preview:")
    print(chunk["text"][:500])
    print("-" * 100)
`````

# 5.1 Check Token Length of Chunks

We chunked by words.

But the model works with tokens.

One word may become one token, multiple tokens, or subword tokens.

So we should check whether chunks are small enough for the QnA model.

`````python
# ============================================================
# Cell 40: Check Token Lengths of Chunks
# ============================================================

def compute_chunk_token_lengths(chunks, tokenizer, sample_question="What is BERT?"):
    """
    Compute tokenized length of question + each chunk.

    This helps us check whether chunks are too long for the model.
    """

    token_length_records = []

    for chunk in chunks:
        encoded = tokenizer(
            sample_question,
            chunk["text"],
            truncation=False,
            return_tensors="pt"
        )

        token_length_records.append({
            "chunk_id": chunk["chunk_id"],
            "page_number": chunk["page_number"],
            "num_words": chunk["num_words"],
            "token_length": encoded["input_ids"].shape[1]
        })

    return pd.DataFrame(token_length_records)


token_lengths_df = compute_chunk_token_lengths(
    chunks=chunks,
    tokenizer=qa_tokenizer,
    sample_question="What does BERT stand for?"
)

token_lengths_df.head()
`````

`````python
# ============================================================
# Cell 41: Token Length Summary
# ============================================================

print("Token length summary:")
print(token_lengths_df["token_length"].describe())

max_allowed_length = 500

too_long_chunks = token_lengths_df[
    token_lengths_df["token_length"] > max_allowed_length
]

print("\nNumber of chunks longer than", max_allowed_length, "tokens:")
print(len(too_long_chunks))

if len(too_long_chunks) > 0:
    print("\nSome chunks may still be truncated.")
    print("Reduce chunk_size if needed.")
else:
    print("\nAll chunks fit within the chosen max_length.")
`````

## Why We Ask the Question Over Every Chunk

A long document cannot be passed directly to the model.

So instead of:

```text

Question + Chunk 1
Question + Chunk 2

Each chunk gives a possible answer.

## Important Teaching Point

This is still not full RAG.

Current method:

```text

Future RAG method:

Then run QnA or generation

So this section is a bridge between:

Document QnA

and

Retrieval-Augmented Generation

`````python
# ============================================================
# Cell 44: Ask a Question Over All Chunks
# ============================================================

def answer_question_over_chunks(
    question,
    chunks,
    tokenizer,
    model,
    device=device,
    top_k=5,
    max_length=384,
    max_answer_length=40,
    min_answer_length=1
):
    """
    Run manual extractive QnA over every chunk.

    Input:
        question: user question
        chunks: list of chunk dictionaries
        tokenizer: Hugging Face tokenizer
        model: Hugging Face QnA model

    Output:
        DataFrame containing ranked candidate answers.
    """

    candidate_answers = []

    for chunk in chunks:
        result = manual_extractive_qa(
            question=question,
            context=chunk["text"],
            tokenizer=tokenizer,
            model=model,
            device=device,
            max_length=max_length,
            max_answer_length=max_answer_length
        )

        answer_text = result["answer"].strip()

        if len(answer_text) >= min_answer_length:
            candidate_answers.append({
                "question": question,
                "answer": answer_text,
                "confidence_score": result["confidence_score"],
                "span_score": result["span_score"],
                "chunk_id": chunk["chunk_id"],
                "page_number": chunk["page_number"],
                "start_word_index": chunk["start_word_index"],
                "end_word_index": chunk["end_word_index"],
                "context_preview": chunk["text"][:300]
            })

    ranked_answers = sorted(
        candidate_answers,
        key=lambda x: x["span_score"],
        reverse=True
    )

    ranked_answers = ranked_answers[:top_k]

    return pd.DataFrame(ranked_answers)
`````

# 6.4 Add a Simple Answer Quality Filter

Sometimes the model extracts unhelpful answers such as:

```text
empty string
very short punctuation
special tokens
irrelevant fragments

We can add a simple filter.

This is not perfect, but it improves the first project version.

`````python
# ============================================================
# Cell 49: Simple Answer Quality Filter
# ============================================================

def is_reasonable_answer(answer):
    """
    Simple heuristic filter for extracted answers.
    """

    answer = answer.strip()

    if len(answer) == 0:
        return False

    if len(answer) < 2:
        return False

    if answer in [".", ",", ";", ":", "-", "–"]:
        return False

    if answer.lower() in ["[cls]", "[sep]", "[pad]"]:
        return False

    return True
`````

`````python
# ============================================================
# Cell 50: Improved QnA Over Chunks with Basic Filtering
# ============================================================

def answer_question_over_chunks_filtered(
    question,
    chunks,
    tokenizer,
    model,
    device=device,
    top_k=5,
    max_length=500,
    max_answer_length=40
):
    """
    Run QnA over chunks and filter unreasonable answers.
    """

    candidate_answers = []

    for chunk in chunks:
        result = manual_extractive_qa(
            question=question,
            context=chunk["text"],
            tokenizer=tokenizer,
            model=model,
            device=device,
            max_length=max_length,
            max_answer_length=max_answer_length
        )

        answer_text = result["answer"].strip()

        if is_reasonable_answer(answer_text):
            candidate_answers.append({
                "question": question,
                "answer": answer_text,
                "confidence_score": result["confidence_score"],
                "span_score": result["span_score"],
                "chunk_id": chunk["chunk_id"],
                "page_number": chunk["page_number"],
                "context_preview": chunk["text"][:300]
            })

    ranked_answers = sorted(
        candidate_answers,
        key=lambda x: x["span_score"],
        reverse=True
    )

    return pd.DataFrame(ranked_answers[:top_k])
`````

# 7.4 Add a Minimum Score Threshold

One simple practical improvement is to reject answers below a threshold.

For example:

```text
If span_score is too low, return:
"No reliable answer found."

This does not fully solve the problem, but it improves system behavior.

`````python
# ============================================================
# Cell 55: Threshold-Based Final Answer Function
# ============================================================

def get_best_answer_with_threshold(
    question,
    chunks,
    tokenizer,
    model,
    device=device,
    span_score_threshold=5.0,
    top_k=5,
    max_length=500,
    max_answer_length=40
):
    """
    Return best answer only if its span score is above a threshold.

    If the score is too low, return a safe fallback response.
    """

    answers_df = answer_question_over_chunks_filtered(
        question=question,
        chunks=chunks,
        tokenizer=tokenizer,
        model=model,
        device=device,
        top_k=top_k,
        max_length=max_length,
        max_answer_length=max_answer_length
    )

    if len(answers_df) == 0:
        return {
            "question": question,
            "answer": "No reliable answer found.",
            "status": "no_candidate_answer",
            "span_score": None,
            "confidence_score": None,
            "page_number": None,
            "chunk_id": None,
            "context_preview": None
        }

    best_answer = answers_df.iloc[0].to_dict()

    if best_answer["span_score"] < span_score_threshold:
        return {
            "question": question,
            "answer": "No reliable answer found in the document.",
            "status": "below_threshold",
            "span_score": best_answer["span_score"],
            "confidence_score": best_answer["confidence_score"],
            "page_number": best_answer["page_number"],
            "chunk_id": best_answer["chunk_id"],
            "context_preview": best_answer["context_preview"]
        }

    return {
        "question": question,
        "answer": best_answer["answer"],
        "status": "answered",
        "span_score": best_answer["span_score"],
        "confidence_score": best_answer["confidence_score"],
        "page_number": best_answer["page_number"],
        "chunk_id": best_answer["chunk_id"],
        "context_preview": best_answer["context_preview"]
    }
`````

## Teaching Note: Thresholds Are Heuristics

A threshold is a practical rule.

It is not a mathematical proof of correctness.

If the threshold is too low:

```text
wrong answers may pass

If the threshold is too high:
```text
correct answers may be rejected


So threshold selection depends on the application.

`````python
# ============================================================
# Cell 56: Test Threshold-Based Answering
# ============================================================

threshold_test_questions = [
    "What does BERT stand for?",
    "What is the author's favorite programming language?",
    "What tasks can BERT be fine-tuned for?"
]

threshold_results = []

for question in threshold_test_questions:
    result = get_best_answer_with_threshold(
        question=question,
        chunks=chunks,
        tokenizer=qa_tokenizer,
        model=qa_model,
        device=device,
        span_score_threshold=5.0,
        top_k=5,
        max_length=500,
        max_answer_length=40
    )

    threshold_results.append(result)

pd.DataFrame(threshold_results)
`````

# 8.4 How This Becomes a RAG Project

Our current system checks every chunk.

Current approach:

```text
Question
→ Run QnA on Chunk 0
→ Run QnA on Chunk 1
→ Run QnA on Chunk 2
→ ...
→ Rank answers

This works for a small PDF.

But for many documents, this becomes slow.

RAG improves this by adding retrieval first.

## RAG Workflow

A RAG system follows this structure:

```text
PDF Documents
        ↓
Text Extraction
        ↓
Chunking
        ↓
Embedding Generation
        ↓
Vector Database
        ↓
User Question
        ↓
Retrieve Top-k Relevant Chunks
        ↓
QnA Model or LLM
        ↓
Grounded Answer

The major improvement is:

Do not read every chunk.
Retrieve the most relevant chunks first.

## Current System vs RAG System

| Aspect | Current Document QnA | RAG |
|---|---|---|
| Chunk handling | Checks every chunk | Retrieves relevant chunks |
| Speed | Slower for large documents | Faster for large collections |
| Intelligence | Depends on QnA model ranking | Depends on retrieval + answering |
| Best use case | Small PDFs | Large PDFs / many documents |
| Next step | Add retrieval | Add generation or fine-tuning |

| Feature | Extractive QnA | LLM-Based QnA |

# Limitations / Teaching Caution

This project is strong as a first document QnA system, but it has limitations.

## 1. PDF Extraction Limitation

PDF text extraction may be noisy.

Tables, formulas, headers, footers, and references may not extract cleanly.

## 2. Scanned PDF Limitation

Scanned PDFs are image-based.

They require OCR before text-based QnA can work.

## 3. Chunking Limitation

Word-based chunking is simple but not semantically perfect.

A better system may use sentence-based, section-based, or semantic chunking.

## 4. Ranking Limitation

Ranking by span score is basic.

Correct answers may appear in top-k but not at rank 1.

## 5. Extractive Limitation

The model can only extract text from context.

It cannot synthesize long explanations like an LLM.

## 6. Confidence Limitation

Confidence score is not correctness.

Human verification through source context is still important.

## 7. Scalability Limitation

Running QnA over every chunk is not scalable.

RAG solves this by adding retrieval before answering.
